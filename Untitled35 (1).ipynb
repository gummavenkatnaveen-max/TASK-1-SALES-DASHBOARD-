```python
import numpy as  np
import pandas as pd
import matplotlib .pyplot as plt
import seaborn as sns
```

```python
df.columns = df.columns.str.capitalize()
```

```python
df['Region'] = df['Region'].fillna(df['Region'].mode()[0])
```

```python
df["Order_date"] = pd.to_datetime(df["Order_date"])
df["Month"] = df["Order_date"].dt.month
```

```python
df
```

```python
df.isnull().sum()
```

```python
region_revenue = (
    df.groupby("Region")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print(region_revenue)
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=region_revenue.index,
    y=region_revenue.values
)

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Monthly_revenue = (
    df.groupby("Month")["Revenue"]
      .sum()
      .sort_values()
)
```

```python
monthly_revenue = Monthly_revenue.sort_index()
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(
    x=monthly_revenue.index,
    y=monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#units sold by regioun and product
region_product_analysis = (
    df.groupby(["Region", "Product"])
      .agg(
          Total_Units=("Units", "sum"),
          Total_Revenue=("Revenue", "sum"),
          Avg_Discount=("Discount_pct", "mean")
      )
      .reset_index()
)

print(region_product_analysis)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=region_product_analysis,
    x="Region",
    y="Total_Units",
    hue="Product"
)

plt.title("Units Sold by Region and Product")
plt.xlabel("Region")
plt.ylabel("Total Units")

plt.tight_layout()
plt.show()
```

```python
#product revenue by  channel
product_channel_customer = df.pivot_table(
    index="Product",
    columns=["Channel", "Customer_type"],
    values="Revenue",
    aggfunc="sum"
)

print(product_channel_customer)
product_channel = (
    df.groupby(["Product", "Channel"])["Revenue"]
      .sum()
      .reset_index()
)
```
